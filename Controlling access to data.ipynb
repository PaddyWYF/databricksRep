```sql
--column filter

create or replace function color_mask(color string) returns string
return case when is_account_group_member('metastore_admins')
then color
else 'REDACTED'
end;

use catalog lakehouse_dev;
use schema datasphere_test;

ALTER TABLE fruit
  alter column color set mask color_mask;
```

```sql
select * from fruit
```

```sql
-- Row filter

create or replace function color_filter(color string) 
return if(is_account_group_member('metastore_admins'),true, color <> 'red' );

```

```sql
alter table fruit set row filter color_filter on (color);
select * from fruit;
```
