```sql
USE CATALOG lakehouse_dev;
-- update schema
USE SCHEMA datasphere_test;
CREATE STREAMING TABLE fruit_streaming
SCHEDULE EVERY 1 HOUR
AS SELECT *, current_timestamp() AS IngestDateTime FROM stream fruit;




```

```sql
select is_account_group_member('admin');


create or replace view fruit_dynamic_view as
select * from fruit 
where 
  case 
  when is_account_group_member('admin') then true
else fruit = 'apple'
end;  


select * from fruit_dynamic_view;



create or replace view fruit_dynamic_view2 as
select Fruit, color, 
  case 
  when is_account_group_member('admin') then amount
else 'No Auth'
end as Amount
from fruit;


select * from fruit_dynamic_view2;
```
